```python
SOURCE_LISTS = {
    "Hallmark": HALLMARK_HYPOXIA,
    "Winter": WINTER_HYPOXIA_METAGENE,
    "Buffa": BUFFA_HYPOXIA_METAGENE,
    "Elvidge": ELVIDGE_HYPOXIA_UP,
    "Fardin": FARDIN_HYPOXIA_11,
    "Toustrup": TOUSTRUP_GENES,
    "NIMRAD": NIMRAD_GENES,
    "BreastScore": BREAST_SCORE_GENES,
}

all_genes = set()
for lst in SOURCE_LISTS.values():
    all_genes.update(lst)

print("Total unique genes across all 8 sources:", len(all_genes))

rows = []
for gene in sorted(all_genes):
    sources = [name for name, lst in SOURCE_LISTS.items() if gene in lst]
    rows.append({"Gene": gene, "N_sources": len(sources), "Sources": ";".join(sources)})

pool_df = pd.DataFrame(rows).sort_values(["N_sources", "Gene"], ascending=[False, True])
pool_df.head(15)
```

```python
core_genes = pool_df[pool_df["N_sources"] >= 5]
print(f"Genes appearing in 5+ of the 8 source lists ({core_genes.shape[0]} genes):")
core_genes[["Gene", "Sources"]]
```

```python
pool_df["In_TCGA_HNSC"] = pool_df["Gene"].apply(lambda g: g in cols)
n_present = pool_df["In_TCGA_HNSC"].sum()
n_total = pool_df.shape[0]
print(f"Coverage: {n_present} / {n_total} genes ({100*n_present/n_total:.1f}%) present in TCGA-HNSC")

missing = pool_df[~pool_df["In_TCGA_HNSC"]]
print(f"\n{missing.shape[0]} genes not found under their current symbol:")
missing[["Gene", "Sources"]]
```

```python
known_aliases = {
    "CCN1": "CYR61", "CCN2": "CTGF", "CCN5": "WISP2",
    "CAVIN1": "PTRF", "CAVIN3": "PRKCDBP", "ERO1A": "ERO1L",
}

for new_symbol, old_symbol in known_aliases.items():
    found = old_symbol in cols
    print(f"{new_symbol} (recorded as '{old_symbol}' in this dataset): {'FOUND' if found else 'still missing'}")

print("\nNote: the remaining ~45 missing genes have not each been individually")
print("alias-checked yet -- a full alias-resolution pass is recommended before")
print("the pool is finally frozen, rather than treating this list as final.")
```
